```python
import pandas as pd

def text_to_excel(input_file, output_file):
    # Read the text file
    with open(input_file, 'r') as file:
        lines = file.readlines()
    
    # Parse lines into a list of lists (each list is a row)
    data = [line.strip().split() for line in lines if line.strip() and line.strip().split()[0]]  # Skip rows where the first entry is null
    
    # Create a DataFrame
    df = pd.DataFrame(data)
    
    # Save to Excel file
    df.to_excel(output_file, index=False, header=False)

# Specify the input text file and output Excel file
input_file = 'input.txt'  # Replace with your input file path
output_file = 'output1.xlsx'  # Replace with your desired output file path

# Convert the text file to Excel
text_to_excel(input_file, output_file)
```
